```python
import pandas as pd
import re
import glob
import os

# ----------------------------
# CONFIG: folder containing load files
# ----------------------------
data_folder = "raw_data"

# ----------------------------

all_records = []
max_load_count = 0  # track max number of loads in any row

# ----------------------------
# STEP 1: Read all txt files
# ----------------------------
for file_path in glob.glob(os.path.join(data_folder, "*.txt")):
    with open(file_path, 'r') as f:
        lines = f.readlines()

    for line in lines:
        # Extract date and hour
        date_str = line[:8]  # first 8 chars
        hour_str = line[8]   # 9th char
        load_values_str = line[9:]  # rest of line

        # Extract all 5-digit numbers from the load string
        loads = [int(x) for x in re.findall(r'\d{5}', load_values_str)]

        # Track max load count for padding later
        if len(loads) > max_load_count:
            max_load_count = len(loads)

        # Append row: [date, hour, loads...]
        all_records.append([date_str, hour_str] + loads)

# ----------------------------
# STEP 2: Pad rows so all have same number of load columns
# ----------------------------
for i in range(len(all_records)):
    row_len = len(all_records[i])
    if row_len < 2 + max_load_count:  # 2 for date & hour
        all_records[i] += [0] * (2 + max_load_count - row_len)

# ----------------------------
# STEP 3: Create DataFrame
# ----------------------------
columns = ['date', 'hour'] + [f'load_{i+1}' for i in range(max_load_count)]
combined_df = pd.DataFrame(all_records, columns=columns)

# ----------------------------
# STEP 4: Clean date and convert to datetime
# ----------------------------
# Keep only the first 8 digits and extract numeric characters
combined_df['date'] = combined_df['date'].astype(str).str[:8].str.extract(r'(\d{8})')[0]

# Convert to datetime (errors='coerce' will mark unparseable dates as NaT)
combined_df['date'] = pd.to_datetime(combined_df['date'], format='%m%d%Y', errors='coerce')

# ----------------------------
# STEP 5: Convert hour to int and create hourly datetime
# ----------------------------
combined_df['hour'] = combined_df['hour'].astype(int)
combined_df['datetime'] = combined_df['date'] + pd.to_timedelta(combined_df['hour'] - 1, unit='h')

# Optional: drop original date and hour columns
combined_df = combined_df.drop(['date', 'hour'], axis=1)

# ----------------------------
# STEP 6: Fill any remaining NaNs (should be none after padding)
# ----------------------------
combined_df = combined_df.fillna(0)

# ----------------------------
# DONE
# ----------------------------
print("Preprocessing complete!")
print("Combined DataFrame shape:", combined_df.shape)
print(combined_df.head())
```

```python
# Go back to combined_df BEFORE melting
# Recreate base date properly

combined_df['base_date'] = combined_df['datetime'].dt.floor('D')

# Melt
load_long = combined_df.melt(
    id_vars=['base_date'],
    value_vars=[col for col in combined_df.columns if col.startswith('load_')],
    var_name='hour_block',
    value_name='load'
)

# Extract hour inside block (0–11)
load_long['hour_in_block'] = load_long['hour_block'].str.extract(r'(\d+)').astype(int) - 1

# Now detect whether row belonged to first or second block
# We determine this using repeating pattern (every 12 loads alternate)

# Reconstruct block index (0 or 1 per day)
load_long['block_index'] = load_long.groupby('base_date').cumcount() // 12

# Final hour = hour_in_block + (block_index * 12)
load_long['final_hour'] = load_long['hour_in_block'] + (load_long['block_index'] * 12)

# Create final datetime
load_long['datetime'] = load_long['base_date'] + pd.to_timedelta(load_long['final_hour'], unit='h')

# Clean
load_long = load_long[['datetime', 'load']].sort_values('datetime').reset_index(drop=True)

print(load_long.shape)
print("Start:", load_long['datetime'].min())
print("End:", load_long['datetime'].max())
```
